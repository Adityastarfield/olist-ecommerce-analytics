# Olist E-Commerce — Load Cleaned Data into SQL Server

Loads all 9 cleaned CSVs (produced by `02_cleaning.ipynb`) into SQL Server via `sqlalchemy` + `pyodbc`, using `to_sql()` for every table. Standardizing on a single, fully-scripted load method rather than mixing in the SSMS Import Flat File wizard keeps this step reproducible directly from this notebook, with no manual GUI steps required.

> **Paths:** this notebook uses relative paths (`../data/raw/...`, `../data/cleaned/...`) and assumes it is opened and run from inside the `01_data_cleaning/` folder, as laid out in this repo. If you move the notebook elsewhere, update the paths accordingly.

```python
import sqlalchemy
import pyodbc
import pandas as pd
```

## Connect to SQL Server

```python
print(pyodbc.drivers())
```

```python
connection_string = "mssql+pyodbc://localhost\\SQLEXPRESS/olist_db?driver=ODBC+Driver+17+for+SQL+Server&trusted_connection=yes"
engine = sqlalchemy.create_engine(connection_string)
```

## Load cleaned CSVs and write to SQL Server

Each table is read from `datasets cleaned/` and written with `if_exists='replace'` so this notebook can be re-run safely from a clean state.

```python
customers_df = pd.read_csv(r"../data/cleaned/customers_clean.csv")
customers_df.to_sql(
    name='customers_clean',
    con=engine,
    if_exists='replace',
    index=False
)
```
